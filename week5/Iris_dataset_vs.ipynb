```python
column_names = [
    'SepalLength',
    'SepalWidth',
    'PetalLength',
    'PetalWidth',
    'Species'
]

df = pd.read_csv(
    'iris.csv',
    header=None,
    names=column_names
)

print(df.head())
```

```python
class_counts = df['Species'].value_counts()
print(class_counts)
```

```python
class_counts.plot(kind='bar')

plt.title('Number of Samples per Iris Species')
plt.xlabel('Species')
plt.ylabel('Number of Samples')
plt.show()
```
